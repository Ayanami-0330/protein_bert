````python
# 功能：加载 Anti-CRISPR 数据并按 sample_id 挂载 310/710/1110 三套 PSSM 特征
train_base_df, test_base_df = load_anticrispr_with_ids(BENCHMARKS_DIR, benchmark_name='anticrispr_binary')

variant_datasets = {}
for variant in FEATURE_VARIANTS:
    parquet_path = f'{WORK_ROOT}/features/pssm_features_{variant}.parquet'
    csv_path = f'{WORK_ROOT}/features/pssm_features_{variant}.csv'
    cache_path = parquet_path if os.path.exists(parquet_path) else csv_path
    feature_df, feature_cols = load_feature_cache(cache_path)

    train_df_v = attach_pssm_features(train_base_df, feature_df, feature_cols)
    test_df_v = attach_pssm_features(test_base_df, feature_df, feature_cols)
    variant_datasets[variant] = (train_df_v, test_df_v, feature_cols)

    print(f'variant={variant} train shape:', train_df_v.shape)
    print(f'variant={variant} test shape:', test_df_v.shape)
    print(f'variant={variant} feature dim:', len(feature_cols))

train_df_310, test_df_310, feature_cols_310 = variant_datasets['310']
train_df_710, test_df_710, feature_cols_710 = variant_datasets['710']
train_df_1110, test_df_1110, feature_cols_1110 = variant_datasets['1110']
````

````python
# 功能：定义二分类评估指标（AUC/AUPRC/F1/MCC/Brier/ECE）及验证集最优阈值搜索
def expected_calibration_error(y_true, y_prob, n_bins=10):
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ids = np.digitize(y_prob, bins) - 1
    ece = 0.0
    n = len(y_true)
    for b in range(n_bins):
        m = ids == b
        if np.any(m):
            conf = float(np.mean(y_prob[m]))
            acc = float(np.mean(y_true[m]))
            ece += (np.sum(m) / n) * abs(acc - conf)
    return float(ece)

def evaluate_binary(y_true, y_prob, thr=0.5):
    y_cls = (y_prob >= thr).astype(int)
    return {
        'AUC': float(roc_auc_score(y_true, y_prob)),
        'AUPRC': float(average_precision_score(y_true, y_prob)),
        'F1': float(f1_score(y_true, y_cls)),
        'MCC': float(matthews_corrcoef(y_true, y_cls)),
        'Brier': float(brier_score_loss(y_true, y_prob)),
        'ECE': float(expected_calibration_error(y_true, y_prob, n_bins=10)),
        'Threshold': float(thr),
    }

def find_best_thr(y_true, y_prob):
    best_thr, best_f1 = 0.5, -1.0
    for thr in np.linspace(0.05, 0.95, 19):
        cur_f1 = f1_score(y_true, (y_prob >= thr).astype(int))
        if cur_f1 > best_f1:
            best_f1, best_thr = cur_f1, float(thr)
    return best_thr
````

````python
# 功能：单次训练评估基线与 PSSM-only 消融（不固定随机种子）
def run_baseline_one_seed(seed=None):
    sub_train, sub_valid = train_test_split(
        train_df_310[['seq', 'label']], test_size=0.1, stratify=train_df_310['label'], random_state=seed
    )

    output_type = OutputType(False, 'binary')
    output_spec = OutputSpec(output_type, [0, 1])

    pretrained_model_generator, input_encoder = load_pretrained_model(
        local_model_dump_dir=f'{PROJECT_ROOT}/proteinbert_models',
        download_model_dump_if_not_exists=True,
        validate_downloading=False,
    )

    mg = FinetuningModelGenerator(
        pretrained_model_generator,
        output_spec=output_spec,
        pretraining_model_manipulation_function=get_model_with_hidden_layers_as_outputs,
        dropout_rate=0.4,
        head_type='two_layer',
        loss_type='bce',
        lr=2e-5,
    )

    finetune(
        mg,
        input_encoder,
        output_spec,
        sub_train['seq'],
        sub_train['label'],
        sub_valid['seq'],
        sub_valid['label'],
        seq_len=512,
        batch_size=8,
        max_epochs_per_stage=8,
        begin_with_frozen_pretrained_layers=True,
        n_final_epochs=0,
    )

    model = mg.create_model(512)
    X_valid = input_encoder.encode_X(sub_valid['seq'].tolist(), 512)
    valid_prob = model.predict(X_valid, batch_size=8, verbose=0).reshape(-1)
    thr = find_best_thr(sub_valid['label'].to_numpy(), valid_prob)

    X_test = input_encoder.encode_X(test_df_310['seq'].tolist(), 512)
    test_prob = model.predict(X_test, batch_size=8, verbose=0).reshape(-1)
    return evaluate_binary(test_df_310['label'].to_numpy(), test_prob, thr=thr)


def run_pssm_only_ablation(train_df, test_df, feature_cols, random_state=None):
    sub_train, sub_valid = train_test_split(
        train_df[['label', *feature_cols]],
        test_size=0.1,
        stratify=train_df['label'],
        random_state=random_state,
    )

    x_train = sub_train[feature_cols].to_numpy(dtype=np.float32)
    x_valid = sub_valid[feature_cols].to_numpy(dtype=np.float32)
    x_test = test_df[feature_cols].to_numpy(dtype=np.float32)

    scaler = StandardScaler()
    x_train = scaler.fit_transform(x_train)
    x_valid = scaler.transform(x_valid)
    x_test = scaler.transform(x_test)

    y_train = sub_train['label'].astype(int).to_numpy()
    y_valid = sub_valid['label'].astype(int).to_numpy()
    y_test = test_df['label'].astype(int).to_numpy()

    clf = LogisticRegression(max_iter=3000, class_weight='balanced')
    clf.fit(x_train, y_train)

    valid_prob = clf.predict_proba(x_valid)[:, 1]
    thr = find_best_thr(y_valid, valid_prob)
    test_prob = clf.predict_proba(x_test)[:, 1]
    return evaluate_binary(y_test, test_prob, thr=thr)
````

````python
# 功能：运行指定实验集合并汇总为 exp_df（不固定随机种子）
cfg = FusionTrainConfig(
    seq_len=512,
    batch_size=8,
    frozen_epochs=6,
    unfrozen_epochs=12,
    frozen_lr=1e-4,
    unfrozen_lr=2e-5,
    pssm_dropout=0.3,
    global_dropout=0.3,
    pssm_hidden_dim=128,
    global_hidden_dim=128,
    global_bottleneck_dim=64,
    fusion_hidden_dim=128,
    use_hidden_global_concat=True,
)


def run_fusion_variant(train_df, test_df, feature_cols):
    pretrained_model_generator, input_encoder = load_pretrained_model(
        local_model_dump_dir=f'{PROJECT_ROOT}/proteinbert_models',
        download_model_dump_if_not_exists=True,
        validate_downloading=False,
    )
    return run_finetune_with_pssm(
        pretrained_model_generator,
        input_encoder,
        train_df,
        test_df,
        feature_cols=feature_cols,
        seed=RANDOM_STATE,
        cfg=cfg,
    )

all_rows = []

all_rows.append({'Exp': 'Baseline_ProteinBERT', **run_baseline_one_seed(seed=RANDOM_STATE)})

all_rows.append({
    'Exp': 'Ablation_RPSSM_110',
    **run_pssm_only_ablation(train_df_310, test_df_310, feature_cols_310[:110], random_state=RANDOM_STATE)
})
all_rows.append({
    'Exp': 'Ablation_RPSSM_310',
    **run_pssm_only_ablation(train_df_310, test_df_310, feature_cols_310, random_state=RANDOM_STATE)
})
all_rows.append({
    'Exp': 'Ablation_RPSSM_710',
    **run_pssm_only_ablation(train_df_710, test_df_710, feature_cols_710, random_state=RANDOM_STATE)
})
all_rows.append({
    'Exp': 'Ablation_RPSSM_1110',
    **run_pssm_only_ablation(train_df_1110, test_df_1110, feature_cols_1110, random_state=RANDOM_STATE)
})

all_rows.append({
    'Exp': 'ProteinBERT_PSSM110',
    **run_fusion_variant(train_df_310, test_df_310, feature_cols_310[:110])
})
all_rows.append({
    'Exp': 'ProteinBERT_PSSM310',
    **run_fusion_variant(train_df_310, test_df_310, feature_cols_310)
})
all_rows.append({
    'Exp': 'ProteinBERT_PSSM710',
    **run_fusion_variant(train_df_710, test_df_710, feature_cols_710)
})
all_rows.append({
    'Exp': 'ProteinBERT_PSSM1110',
    **run_fusion_variant(train_df_1110, test_df_1110, feature_cols_1110)
})

exp_df = pd.DataFrame(all_rows)
exp_df
````

````python
# 功能：按实验组汇总各指标均值（不再统计 std），按 AUPRC 排序
summary = exp_df.groupby('Exp')[['AUC','AUPRC','F1','MCC','Brier','ECE']].mean()
summary = summary.sort_values('AUPRC', ascending=False)
summary
````

````python
# 功能：将实验结果明细与汇总表保存到 data 专用目录（不写 exp_results.csv/exp_summary.csv）
os.makedirs(DEMO_RESULTS_DIR, exist_ok=True)
report_path = f'{DEMO_RESULTS_DIR}/anticrispr_demo_results.csv'
summary_path = f'{DEMO_RESULTS_DIR}/anticrispr_demo_summary.csv'
exp_df.to_csv(report_path, index=False)
summary.to_csv(summary_path)
print('saved:', report_path)
print('saved:', summary_path)
````
